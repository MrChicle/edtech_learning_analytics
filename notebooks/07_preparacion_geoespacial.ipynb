# Preparación Geoespacial para ArcGIS Online

Este notebook prepara los datos del proyecto EdTech Learning Analytics para su análisis geoespacial en ArcGIS Online.

## Objetivo

Incorporar una dimensión geográfica sintética al nivel estudiante y generar archivos compatibles con ArcGIS Online, manteniendo sin cambios las métricas académicas originales.

## Carga de datos

Se utilizan los datasets de estudiantes, datos limpios y datos publicables generados en etapas anteriores del proyecto.

## Validación de identificadores

Se verifica la relación entre `student_id` y `student_id_hash` para mantener la trazabilidad y utilizar únicamente el identificador protegido en los archivos geoespaciales.

## Agregación de métricas

Las interacciones se agrupan a nivel estudiante para evitar que una mayor cantidad de registros genere un peso espacial artificial.

## Indicadores por estudiante

Se calculan métricas de rendimiento, finalización, participación, duración, solicitudes de ayuda e interacción con compañeros.

## Dimensión geográfica sintética

Debido a que el dataset original no contiene ubicaciones suficientes, se incorpora una dimensión geográfica sintética basada en regiones, estados y ciudades de Estados Unidos.

## Aclaración metodológica
> Las ubicaciones generadas tienen fines exclusivamente académicos y no representan la localización real de los estudiantes.

## Regiones geográficas

La distribución utiliza cuatro regiones de Estados Unidos: Northeast, Midwest, South y West.

## Asignación geográfica

Cada estudiante recibe una ubicación fija y reproducible. Los valores académicos originales no son modificados durante este proceso.

## Generación de coordenadas

Se generan coordenadas de latitud y longitud alrededor de ciudades de referencia para permitir la representación de puntos en ArcGIS Online.

## Sistema de referencia

Las coordenadas se generan en WGS 84, sistema compatible con ArcGIS Online para datos de latitud y longitud.

## Dataset para ArcGIS Online

Se genera un archivo con una fila por estudiante que combina ubicación geográfica y métricas académicas agregadas.

## Resumen por estado

También se genera un dataset agregado por estado para facilitar mapas temáticos y comparaciones territoriales.

## Validación de resultados

Se verifican identificadores únicos, coordenadas faltantes, rangos geográficos y distribución de estudiantes por región y estado.

## Archivos generados

- `edtech_students_geo.csv`
- `edtech_student_geo_analysis.csv`
- `edtech_state_geo_summary.csv`

## Uso en ArcGIS Online

El archivo `edtech_student_geo_analysis.csv` será utilizado como fuente principal para mapas temáticos y análisis de patrones espaciales.

## Conclusión

La preparación realizada permite incorporar análisis geoespacial al proyecto sin alterar los datos académicos originales y manteniendo la trazabilidad de las etapas anteriores.

### Mapa 1. Distribución geográfica de estudiantes

Se utilizó clustering en ArcGIS Online para visualizar la concentración espacial de los estudiantes del escenario geográfico sintético.

El mapa permite identificar zonas con mayor densidad de estudiantes, principalmente en áreas metropolitanas distribuidas entre las regiones West, South, Midwest y Northeast.

El clustering representa agrupamiento visual y no implica significancia estadística.

### Mapa 2. Rendimiento académico

Se utilizó `average_score` como variable temática para representar espacialmente el rendimiento académico promedio de cada estudiante.

La simbología permite identificar diferencias territoriales entre valores altos y bajos dentro del escenario geográfico sintético.

El mapa es descriptivo y todavía no representa significancia estadística.

### Mapa 3. Tasa promedio de finalización

Se utilizó `average_completion_rate` para representar espacialmente el nivel promedio de finalización de actividades por estudiante.

La escala de color permite identificar zonas con valores relativamente altos y bajos dentro del escenario geográfico sintético.

El mapa es descriptivo y no implica todavía significancia estadística.

### Mapa 4. Heat Map de distribución estudiantil

Se generó un mapa de calor en ArcGIS Online para visualizar las áreas con mayor concentración espacial de estudiantes.

Las zonas de mayor intensidad representan una mayor densidad de puntos dentro del escenario geográfico sintético.

El Heat Map es una herramienta descriptiva de densidad y no representa un análisis de puntos calientes estadísticamente significativos.

### Hallazgo geoespacial

**Evidencia:** El Heat Map muestra concentraciones de estudiantes en diversas áreas metropolitanas, principalmente en sectores de las regiones West, Northeast y South.

**Contexto:** La distribución geográfica utilizada es sintética y fue incorporada con fines académicos para aplicar técnicas de análisis espacial.

**Interpretación:** Las áreas de mayor intensidad representan zonas con una concentración relativamente mayor de estudiantes dentro del escenario generado.

**Decisión posible:** Utilizar estas zonas como referencia para segmentar análisis académicos y comparar indicadores como rendimiento, finalización y participación.

**Limitación:** La concentración observada depende de la geografía sintética incorporada al dataset y no representa patrones reales de una población estudiantil.
